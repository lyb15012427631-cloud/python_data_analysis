## 练习5：两种广播调整（20分）

以下为独立假设，全部在原`production`基础上分别计算，不连续叠加，也不把模拟调整当作真实变化。

①每年小麦+1、稻谷−1、玉米+2万吨：创建一维`crop_adjustment`，利用广播得到`adjusted_by_crop`。

②2021全部品种+1、2022全部品种+2、2023全部品种+3万吨：把`[1, 2, 3]`转换为`(3, 1)`列向量`year_adjustment`，广播得到`adjusted_by_year`。可另选一种列向量方法复核。

先预测两个结果再运行，输出调整向量形状并解释哪个按列调整、哪个按行调整。特别说明为什么第二种需要列向量：这里原表恰好是3×3，一维长度3虽然可直接相加，实际会按列调整，结果可能错误却不报错。

```python
crop_adjustment = np.array([1., -1., 2.])
adjusted_by_crop = production + crop_adjustment
year_adjustment = np.array([1., 2., 3.]).reshape(3, 1)
adjusted_by_year = production + year_adjustment
print('按粮食调整向量形状：', crop_adjustment.shape)
print('按粮食调整：\n', adjusted_by_crop)
print('按年份调整向量形状：', year_adjustment.shape)
print('按年份调整：\n', adjusted_by_year)
```

**参考解释：** 形状(3,)按最后一维匹配，每行都加[1,-1,2]，因此按粮食列调整；(3,1)在每行的3列扩展，因此按年份行调整。3×3表直接加长度3的一维数组不会报错，却不符合按年份调整的业务要求。

## 拓展A：两种乘法是否相同（选做，不计基础分）

用新的数学演示数组`A = [[1, 2], [3, 4]]`、`B = [[2, 0], [1, 3]]`。分别计算逐元素乘法`element_product`、矩阵乘法`matrix_product`，再用`np.matmul`复核矩阵乘法。

先手算左上角结果，再运行。解释两个左上角为什么不同。本题数组仅用于数学演示，无粮食单位。

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 3]])
element_product = A * B
matrix_product = A @ B
print('逐元素乘法：\n', element_product)
print('矩阵乘法：\n', matrix_product)
print('matmul结果：\n', np.matmul(A, B))
```

**参考解释：** 逐元素结果[[2,0],[3,12]]，矩阵乘法结果[[4,6],[10,12]]。左上角逐元素为1×2=2，矩阵乘法为1×2+2×1=4。

## 拓展B：转置与逆矩阵（选做，不计基础分）

创建数学演示方阵`C = [[2., 1.], [1., 3.]]`。①用两种方法转置；②用`np.linalg.inv`求逆；③用`np.linalg.solve(C, np.eye(2))`求得同一逆矩阵；④用教师给定的`np.allclose`检查两种结果以及`C @ C的逆`是否近似单位矩阵。

解释为何浮点结果用近似比较，而不能只凭屏幕看起来相同。本题不要求研究复数；这些值没有粮食业务单位。

```python
C = np.array([[2., 1.], [1., 3.]])
transpose_one = C.T
transpose_two = np.transpose(C)
C_inverse = np.linalg.inv(C)
C_inverse_solve = np.linalg.solve(C, np.eye(2))
print('转置1：\n', transpose_one)
print('转置2：\n', transpose_two)
print('逆矩阵：\n', C_inverse)
print('两种求逆结果近似相同：', np.allclose(C_inverse, C_inverse_solve))
print('乘回接近单位矩阵：', np.allclose(C @ C_inverse, np.eye(2)))
```

**参考解释：** 该方阵可逆，逆为[[0.6,-0.2],[-0.2,0.4]]。浮点运算可能有舍入误差，因此用allclose检查容差内一致；屏幕格式会舍入，不能代替数值检查。本例C对称，转置等于原矩阵，但不能据此断言任何实数矩阵都等于自身转置。
