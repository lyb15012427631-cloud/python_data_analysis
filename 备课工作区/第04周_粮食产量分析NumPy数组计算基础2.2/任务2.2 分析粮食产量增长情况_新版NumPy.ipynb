### 补充：求逆的另一种方式与方程求解（选做）

浮点结果用`np.allclose`检查。奇异矩阵不可逆；病态矩阵即使不报错，求逆也可能不准确。`pinv`计算伪逆，不能当作普通逆矩阵的无条件等价替换。

```python
identity_matrix = np.eye(matr3.shape[0])
# A X = I的解X就是A的逆；教学展示同一目标的另一方法
inverse_by_solve = np.linalg.solve(matr3, identity_matrix)
print('solve求得的逆矩阵：\n', inverse_by_solve)
print('两种求逆方法近似相同：', np.allclose(inverse_matrix, inverse_by_solve))
print('A @ A的逆近似为单位矩阵：', np.allclose(matr3 @ inverse_matrix, identity_matrix))
# 如果目标只是解A x = b，优先直接solve，避免先计算完整逆矩阵
right_hand_side = np.array([1., 2., 3.])
solution = np.linalg.solve(matr3, right_hand_side)
print('方程解x：', solution)
print('代回A x = b成立：', np.allclose(matr3 @ solution, right_hand_side))
```

## 3. 数组逐元素运算（原代码15—19，基础）

**本部分标识：算术ufunc（通用函数）及其运算符写法。**

ufunc是NumPy的一类函数对象。这里先理解普通逐元素ufunc：把同一种运算应用到数组各个位置，支持广播和数据类型处理；不必由我们逐项写Python循环。常见函数有`np.add`、`np.subtract`、`np.multiply`、`np.divide`、`np.power`。

以本节`x=[1, 2, 3]`、`y=[4, 5, 6]`为例，`np.add(x, y)`依次计算`1+4`、`2+5`、`3+6`，得到`[5, 7, 9]`，形状仍为`(3,)`。对本例普通ndarray，`x + y`是简洁写法，与`np.add(x, y)`结果相同。**说“这些运算属于ufunc内容”，是指数组运算对应ufunc；运算符号本身不是函数对象。** 对普通Python列表写`+`是列表拼接，不是数组逐元素加法。

| 数组表达式 | 对应的ufunc函数写法 | 逐元素含义 |
|---|---|---|
| `x + y` | `np.add(x, y)` | 对应位置相加 |
| `x - y` | `np.subtract(x, y)` | 对应位置相减 |
| `x * y` | `np.multiply(x, y)` | 对应位置相乘 |
| `x / y` | `np.divide(x, y)` | 对应位置相除 |
| `x ** y` | `np.power(x, y)` | 每个x元素取相应y元素次幂 |

这些都是二元ufunc，即接收两个输入、产生一个输出；“二元”指输入数量，不是数组维度。除同形状数组外，ufunc还能通过广播处理兼容形状，第5、6部分会继续演示。

二维数组同样遵循逐元素规则；`*`不能代替数学矩阵乘法，数组`**`不能代替数学矩阵幂。整数幂还需注意整数类型可表示的范围。

拓展辨别：前面演示的`np.matmul`也是ufunc，但属于**广义ufunc**，以矩阵的核心维度计算乘积，并不是普通逐元素相乘。矩阵部分保持原样，本节只补充这个分类。

依据：[NumPy官方ufunc说明](https://numpy.org/doc/stable/reference/ufuncs.html)。

```python
# 算术ufunc：+对应np.add，逐元素相加
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
print('数组相加结果为：', x + y)  # 数组相加
```

```python
# 算术ufunc：-对应np.subtract，逐元素相减
print('数组相减结果为：', x - y)  # 数组相减
```

```python
# 算术ufunc：*对应np.multiply，逐元素相乘
print('数组相乘结果为：', x * y)  # 数组相乘
```

```python
# 算术ufunc：/对应np.divide，逐元素相除
print('数组相除结果为：', x / y)  # 数组相除
```

```python
# 算术ufunc：**对应本例np.power的逐元素幂
print('数组幂运算结果为：', x ** y)  # 数组幂运算
```

### 补充：显式调用算术ufunc（选做）

下面五个函数就是ufunc对象，分别对应前面`+`、`-`、`*`、`/`、`**`的数组计算。本例结果相同，两种写法都有效；函数写法更方便识别ufunc名称。

先预测输出，再与前面的运算符结果对照。数组`**`是逐元素幂；数学矩阵整数次幂另用`np.linalg.matrix_power(A, n)`，后者不是本例普通逐元素ufunc。

```python
# 显式调用五个算术ufunc，结果与前面的数组运算符写法一致
print('add：', np.add(x, y))
print('subtract：', np.subtract(x, y))
print('multiply：', np.multiply(x, y))
print('divide：', np.divide(x, y))
print('power：', np.power(x, y))
```

```python
test_matrix = np.matrix([[1, 2, 3], [2, 3, 4], [3, 4 ,5]])
print("数学矩阵幂计算\n", test_matrix ** 2)
test_array = np.array([[1, 2, 3], [2, 3, 4], [3, 4 ,5]])
print("逐元素幂计算\n", test_array ** 2)
try:
    print(np.linalg.matrix_power(test_array, test_array))
except TypeError as error:
    print(type(error).__name__, ":", error)

print("对ndarray使用数学矩阵幂运算\n", np.linalg.matrix_power(test_array, 2))
print("对于矩阵和数组而言，幂运算是不一样的，数组的幂运算使用的是ufunc，逐元素进行运算，而矩阵的幂运算遵守数学矩阵幂运算的规则")
print(np.array_equal(test_matrix ** 2, test_array ** 2))
print(np.array_equal(test_matrix ** 2, np.linalg.matrix_power(test_array, 2)))
```

### 补充：如何确认一个函数是ufunc（选做）

使用`isinstance(函数, np.ufunc)`直接检查对象类型。检查函数本身时不写调用括号，例如检查`np.add`，不是检查`np.add(x, y)`的数组结果。

`nin`和`nout`分别表示输入、输出数量；本节算术函数是二元ufunc。分类确认后，回到算术、比较与广播理解其实际作用；无需记忆所有对象属性。

```python
ufunc_checks = {
    'np.add（算术）': np.add,
    'np.less（比较）': np.less,
    'np.logical_and（逻辑）': np.logical_and,
    'np.matmul（广义ufunc）': np.matmul,
    'np.all（归约函数）': np.all,
    'np.any（归约函数）': np.any,
    'np.reshape（形状调整）': np.reshape,
    'np.broadcast_to（广播视图）': np.broadcast_to,
}
for name, function in ufunc_checks.items():
    print(name, '是否为ufunc：', isinstance(function, np.ufunc))
print('np.add的输入数量、输出数量：', np.add.nin, np.add.nout)
```

## 4. 比较与逻辑判断（原代码20—27，基础）

**本部分标识：比较ufunc → 布尔数组 → 逐元素逻辑ufunc或归约判断。**

| 数组比较 | 对应ufunc | 返回值 |
|---|---|---|
| `x < y` | `np.less(x, y)` | 各位置是否小于 |
| `x > y` | `np.greater(x, y)` | 各位置是否大于 |
| `x == y` | `np.equal(x, y)` | 各位置是否相等 |
| `x >= y` | `np.greater_equal(x, y)` | 各位置是否大于或等于 |
| `x <= y` | `np.less_equal(x, y)` | 各位置是否小于或等于 |
| `x != y` | `np.not_equal(x, y)` | 各位置是否不等 |

比较ufunc逐元素计算，得到同一位置对应的`True/False`。本例`np.equal([1, 3, 5], [2, 3, 4])`返回`[False, True, False]`。

后续有两种不同的处理：

- **逐元素逻辑ufunc**：`np.logical_and`、`np.logical_or`、`np.logical_not`，继续在各个位置判断多个条件，不把整个数组合成一个结论。
- **归约函数**：`np.all`表示全部成立，`np.any`表示至少一个成立；默认把所有元素归约为一个布尔结论，也可指定`axis`。**np.all/np.any本身不是ufunc对象**，不能因为它们出现在NumPy逻辑判断章节就把它们都标成ufunc。

因此，`np.all(x == y)`包含两个步骤：先通过比较ufunc产生布尔数组，再归约为总体判断。不要用Python的`and`/`or`直接连接多个布尔数组。

```python
# 比较ufunc：<对应np.less，逐元素返回布尔值
x = np.array([1, 3, 5])
y = np.array([2, 3, 4])
print('数组比较结果为：', x < y)
```

```python
# 比较ufunc：>对应np.greater
print('数组比较结果为：', x > y)
```

```python
# 比较ufunc：==对应np.equal
print('数组比较结果为：', x == y)
```

```python
# 比较ufunc：>=对应np.greater_equal
print('数组比较结果为：', x >= y)
```

```python
# 比较ufunc：<=对应np.less_equal
print('数组比较结果为：', x <= y)
```

```python
# 比较ufunc：!=对应np.not_equal
print('数组比较结果为：', x != y)
```

```python
# 先用比较ufunc得到布尔数组，再用all归约；np.all本身不是ufunc
print('数组逻辑运算结果为：', np.all(x == y))
```

```python
# 先用比较ufunc得到布尔数组，再用any归约；np.any本身不是ufunc
print('数组逻辑运算结果为：', np.any(x == y))
```

### 补充：比较ufunc、逻辑ufunc与all/any（选做）

下面先显式调用六个比较ufunc，再演示布尔数组`.all()`/`.any()`方法，最后用逻辑ufunc组合范围条件。范围筛选可用`np.logical_and`或带括号的`&`；在**布尔数组**上二者结果相同。

严格区分：`&`对应按位ufunc `np.bitwise_and`，`|`对应`np.bitwise_or`，布尔输入时可表示逐元素逻辑与/或；整数输入时按位运算，不能泛化为与logical_and/logical_or完全相同。浮点数近似比较可用`np.isclose`，它本身不是ufunc，也不等价于精确的`==`。

```python
# 六个比较函数是ufunc；all/any是归约；logical_and是逐元素逻辑ufunc
x = np.array([1, 3, 5])
y = np.array([2, 3, 4])
comparisons = {
    '小于': np.less(x, y),
    '大于': np.greater(x, y),
    '等于': np.equal(x, y),
    '大于等于': np.greater_equal(x, y),
    '小于等于': np.less_equal(x, y),
    '不等于': np.not_equal(x, y),
}
for name, result in comparisons.items():
    print(name, result)
print('数组方法all：', (x == y).all())
print('数组方法any：', (x == y).any())
# 多条件逐元素组合，两种写法得到相同的布尔数组
print('logical_and：', np.logical_and(x >= 3, y <= 3))
print('带括号的&：', (x >= 3) & (y <= 3))

print("不能直接使用python中的逻辑判断进行多元素的布尔值判断".center(100, "="))
try:
    print((x >= 3) and (y <= 3))
except Exception as e:
    print(type(e).__name__, ":", e)
print("不能直接使用python中的逻辑判断进行多元素的布尔值判断".center(100, "="))

print("如果使用python中的逻辑判断进行数组逐元素判断，可以使用for循环进行：")

bool_list = []
for index, (i, j) in enumerate(zip(x, y)):
    print(f"进行位置{index}的元素判断，结果是：", end='')
    result = bool(i >= 3 and j<= 3)
    bool_list.append(result)
    print(result)
print("最终的结果是：", bool_list)

print('logical_or（至少一个逐元素条件成立）：', np.logical_or(x >= 3, y <= 3))
print('logical_not（逐元素取反）：', np.logical_not(x == y))
```

### 补充：ufunc也有归约方法reduce（选做）

普通调用如`np.add(x, y)`逐元素生成数组；同一个ufunc还可调用`.reduce()`，沿指定轴反复运算，汇总多个元素。下面只观察其与本节all/any的联系：对于本例布尔数组，`np.logical_and.reduce(..., axis=None)`与`np.all(...)`结果相同，`np.logical_or.reduce(..., axis=None)`与`np.any(...)`结果相同。

**等价结果不代表对象类型相同**：logical_and/logical_or是ufunc，reduce是它们的方法，all/any是归约函数。这里显式写`axis=None`归约全部元素，避免混淆`ufunc.reduce`默认axis=0与all/any默认归约所有元素的差异。

依据：[NumPy官方reduce说明](https://numpy.org/doc/stable/reference/generated/numpy.ufunc.reduce.html)。

```python
equal_mask = np.equal(x, y)  # 比较ufunc先生成布尔数组
print('布尔数组：', equal_mask)
print('逐元素逻辑结果：', np.logical_and(equal_mask, equal_mask))
print('logical_and.reduce全部归约：', np.logical_and.reduce(equal_mask, axis=0))
print('np.all全部归约：', np.all(equal_mask))
print('logical_or.reduce全部归约：', np.logical_or.reduce(equal_mask, axis=0))
print('np.any全部归约：', np.any(equal_mask))
```

## 5. 广播：按行相加（原代码28—32，基础）

**本部分标识：算术ufunc `np.add`的广播能力。**

`arr1 + arr2`对应本例`np.add(arr1, arr2)`。形状`(4, 3)`与`(3,)`从末尾维度对齐：可把后者理解为`(1, 3)`，长度1的维度能扩展为4；于是每一行加上同一个长度3的向量。这相当于给三个指标分别加上固定调整值。

教学理解顺序：①查看形状；②从末尾比较，每一维要相等或有一个为1，缺少的前导维度视为1；③形成兼容的输出形状；④ufunc逐元素运算。广播通常不需要把小数组真正复制4份。

广播是形状匹配规则，**不是另一个ufunc名称**；ufunc使用这套规则完成逐元素计算。

```python
arr1 = np.array([[0, 0, 0], [1, 1, 1], [2, 2, 2], [3, 3, 3]])
print('创建的数组arr1为：\n', arr1)
```

```python
print('数组arr1的形状为：', arr1.shape)
```

```python
arr2 = np.array([1, 2, 3])
print('创建的数组arr2为：', arr2)
```

```python
print('数组arr2的形状为：', arr2.shape)
```

```python
# np.add是ufunc：先按(4, 3)与(3,)的规则广播，再逐元素相加
print('数组相加结果为：\n', arr1 + arr2)
```

### 补充：ufunc广播的函数写法与观察方法（选做）

`np.add`是ufunc，下面两种写法仍计算同一个加法。`reshape`用于调整形状，`np.broadcast_to`用于观察广播后的只读视图；**它们本身不是ufunc**。实际加法直接广播即可，不必先重复复制数据。

```python
# np.add是ufunc；后面的reshape和broadcast_to不是ufunc
print('arr1', arr1)
print('arr2', arr2)
print('np.add同样自动广播：\n', np.add(arr1, arr2))
row_adjustment = arr2.reshape(1, -1)
print('显式行向量：', row_adjustment)
print('显式行向量形状：', row_adjustment.shape)
print('显式行向量加法：\n', arr1 + row_adjustment)
print('展示广播后的向量：\n', np.broadcast_to(arr2, arr1.shape))
```

## 6. 广播：按列相加（原代码33—37，基础）

**本部分标识：同一个算术ufunc `np.add`，配合列向量广播。**

形状`(4, 3)`与`(4, 1)`可以广播，列向量的每个值在该行3列使用，每一行加上该行对应的调整值。长度4的一维数组形状为`(4,)`，末尾4与3不兼容，不能直接与`(4, 3)`相加。

前面的按行广播和本节按列广播使用的都是`np.add`；区别来自输入形状，而不是换了加法函数。`reshape`仅组织行列，真正进行逐元素加法的是ufunc。

```python
arr1 = np.array([[0, 0, 0], [1, 1, 1], [2, 2, 2], [3, 3, 3]])
print('创建的数组arr1为：\n', arr1)
```

```python
arr2 = np.array([1, 2, 3, 4]).reshape((4, 1))
print('创建的数组arr2为：\n', arr2)
```

```python
print('数组arr2的形状为：', arr2.shape)
```

```python
# np.add是ufunc：先按(4, 3)与(4, 1)的规则广播，再逐元素相加
print('数组相加结果为：\n', arr1 + arr2)
```

### 补充：列向量的其他创建方法（选做）

`reshape`中的`-1`表示自动推断长度。`np.atleast_2d(values)`得到`(1, 4)`行向量；需要列向量时应再转置为`(4, 1)`。

这里的reshape、newaxis/None索引、expand_dims是形状调整或索引方法，本身不是ufunc；末尾的`np.add(arr1, arr2)`才是使用广播的算术ufunc。

```python
# 先调整列向量形状，再由np.add执行支持广播的逐元素计算
print("arr1:\n", arr1)
print("arr2:\n", arr2)
values = np.array([1, 2, 3, 4])
column_methods = {
    'reshape函数': np.reshape(values, (-1, 1)),
    'newaxis索引': values[:, np.newaxis],
    'None索引': values[:, None],
    'expand_dims函数': np.expand_dims(values, axis=1),
}
for method, column in column_methods.items():
    print(method, '→形状', column.shape, '结果相同：', np.array_equal(arr1 + column, arr1 + arr2))
print('函数形式按列广播相加：\n', np.add(arr1, arr2))
```
