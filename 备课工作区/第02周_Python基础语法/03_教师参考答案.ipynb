## 1. 支出分档（基础，15分）

定义classify_expense(amount)，规则：负数返回"待核对"；0≤金额<100返回"小额"；100≤金额<500返回"中额"；金额≥500返回"大额"。输入为数值。测试-1、0、99、100、499、500。


```python
def classify_expense(amount):
    if amount < 0:
        return "待核对"
    elif amount < 100:
        return "小额"
    elif amount < 500:
        return "中额"
    else:
        return "大额"

for amount in [-1, 0, 99, 100, 499, 500]:
    print(amount, classify_expense(amount))
```

**核对与讲评：** 依次为：待核对、小额、小额、中额、中额、大额。边界100和500不能落入前一档。


## 2. 循环统计（基础，15分）

对[35, 120, 0, 68, 250]使用for统计总支出和至少100元的笔数。此题先不用sum，体会累加和计数；再用sum核对。


```python
expenses = [35, 120, 0, 68, 250]
total = 0
large_count = 0
for expense in expenses:
    total = total + expense
    if expense >= 100:
        large_count = large_count + 1
print(total, large_count)
print("用sum核对：", sum(expenses))
```

**核对与讲评：** 总支出473元，至少100元共2笔。初始化应在循环前，最终输出应在循环后。


## 3. range重复任务（基础，10分）

使用range，每月存入200元，连续3个月，逐月输出累计金额。无利息，不是复利计算。


```python
total_saved = 0
for month in range(1, 4):
    total_saved = total_saved + 200
    print(month, total_saved)
```

**核对与讲评：** 输出1 200、2 400、3 600；range的停止值不包含在内。


## 4. while达标（基础，10分）

初始余额0，每次存入150元，达到至少500元停止。输出次数和最终余额。将目标设为0再试一次。


```python
balance = 0
target = 500
months = 0
while balance < target:
    balance = balance + 150
    months = months + 1
print(months, balance)
```

**核对与讲评：** 目标500时4次、600元；目标0时0次、0元。最后一次可以超过目标。


## 5. 平均值函数（基础，15分）

定义average_expense(values)，输入为非负数值列表，返回平均值，空列表返回None。调用时再使用round展示，不在函数中提前四舍五入。


```python
def average_expense(values):
    if len(values) == 0:
        return None
    return sum(values) / len(values)

print(average_expense([35, 120, 0, 68, 250]))
print(average_expense([]))
```

**核对与讲评：** 正常列表返回94.6；空列表返回None。0元仍是一笔有效记录。

## 7. 综合分析（基础，25分）

定义expense_report(raw_values, budget)。输入数值列表；本题非负为有效，负数为待核对。函数打印有效笔数、待核对笔数、总额、均值、最小最大、至少100元的记录，以及是否超过预算。无有效数据时打印“暂无有效数据”，不计算均值和极值。使用[60, -5, 100, 40, 200, 0]、预算350元。最后写一句业务解释。不要修改原列表。


```python
def expense_report(raw_values, budget):
    valid_values = []
    large_values = []
    invalid_count = 0
    for value in raw_values:
        if value < 0:
            invalid_count = invalid_count + 1
        else:
            valid_values.append(value)
            if value >= 100:
                large_values.append(value)
    print("有效笔数：", len(valid_values))
    print("待核对笔数：", invalid_count)
    if len(valid_values) == 0:
        print("暂无有效数据")
        return
    total = sum(valid_values)
    print("总额：", total)
    print("均值：", round(total / len(valid_values), 2))
    print("最小最大：", min(valid_values), max(valid_values))
    print("至少100元：", large_values)
    if total > budget:
        print("超预算：", total - budget)
    else:
        print("预算内剩余：", budget - total)

raw_expenses = [60, -5, 100, 40, 200, 0]
expense_report(raw_expenses, 350)
expense_report([], 350)
print("原始列表：", raw_expenses)
```

**核对与讲评：** 有效5笔、待核对1笔，总额400元、均值80元、最小0、最大200，至少100元为[100, 200]，超预算50元。解释需说明负数被排除只是本题约定，0元计入分母。


## 8. 数据配对（拓展，不计基础分）

用enumerate显示每笔的序号；再用zip配对["餐饮", "交通", "学习"]和[35, 12, 80]。先检查两组长度，长度不一致时提示核对，不能静默漏记录。


```python
categories = ["餐饮", "交通", "学习"]
expenses = [35, 12, 80]
for number, expense in enumerate(expenses, start=1):
    print(number, expense)
if len(categories) != len(expenses):
    print("长度不一致，请核对")
else:
    for category, expense in zip(categories, expenses):
        print(category, expense)
```

**核对与讲评：** 序号从1开始，配对为餐饮35、交通12、学习80；删除一个金额后应提示长度不一致。
